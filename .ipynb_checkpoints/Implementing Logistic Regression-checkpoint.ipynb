```python
import pandas as pd
import Log_Reg
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
diabetes_data = pd.read_csv("diabetes.csv")
```

```python
diabetes_data.head()
```

```python
diabetes_data.isnull().sum()
```

```python
diabetes_data.shape
```

```python
X = diabetes_data.drop(['Outcome'], axis = 1)
Y = diabetes_data['Outcome']
```

```python
scalar = StandardScaler()
```

```python
standardized_data = scalar.fit_transform(X)
```

```python
X = standardized_data
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y, random_state = 2)
```

```python
print(X_train, X_test, Y_train, Y_test)
```

```python
model = Log_Reg.Logistic_Regression(learning_rate = 0.01, iterations = 1000)
```

```python
model.fit(X_train, Y_train)
```

```python
# Accuracy score on training data
X_train_prediction = model.predict_classes(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
```

```python
print(f'Accuracy score on training data = {training_data_accuracy}')
```

```python
#Accuracy score on test data
X_test_prediction = model.predict_classes(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
```

```python
print(f'Accuracy score on test data = {test_data_accuracy}')
```

```python
#making a predictive system
import numpy as np

# Input data
input_data = (10,115,0,0,0,35.3,0.134,29)
input_data_as_numpy_array = np.asarray(input_data)

# Reshape for a single sample
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

# Standardize using the fitted scaler
std_data = scalar.transform(input_data_reshaped)

# Predict class (0 or 1)
prediction = model.predict_classes(std_data)
print("Prediction:", prediction)

if prediction[0] == 1:
    print('The person is diabetic')
else:
    print('The person is non-diabetic')
```
